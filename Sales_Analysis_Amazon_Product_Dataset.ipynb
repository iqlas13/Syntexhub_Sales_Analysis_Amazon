## 🧹 Data Cleaning and Preprocessing

Real-world datasets often contain inconsistencies such as missing values, incorrect data types, and unwanted symbols.  
Before performing sales analysis, it is important to clean and preprocess the dataset to ensure accurate results.

In this step, we focus on:
- Handling missing values
- Converting price columns to numeric format
- Cleaning discount percentage values
- Preparing rating-related columns for analysis

```python
# Check for missing values
df.isnull().sum()
```

```python
import pandas as pd

# Re-load the dataset to ensure df is a DataFrame
df = pd.read_csv("/mnt/data/amazon.csv")

# Drop rows with missing values
df = df.dropna()
# Verify missing values are handled
df.isnull().sum()
```

```python
# Remove missing values safely
df = df.dropna().copy()
```

```python
# Clean actual_price column
df.loc[:, 'actual_price'] = (
    df['actual_price']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)

# Clean discounted_price column
df.loc[:, 'discounted_price'] = (
    df['discounted_price']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)
```

```python
df[['actual_price', 'discounted_price']].head()
```

### Price Column Cleaning

The currency symbols and commas present in the price columns were removed to convert the data into numerical format.  
To avoid chained assignment issues in Pandas, an explicit copy of the DataFrame was created and values were updated using `.loc`.

This ensures safe and accurate data preprocessing for further analysis.


```python
# Convert rating to numeric
df.loc[:, 'rating'] = pd.to_numeric(df['rating'], errors='coerce')

# Convert rating_count to numeric
df.loc[:, 'rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')
```

```python
df = df.dropna().copy()
```

```python
df[['rating', 'rating_count']].head()
```

### Rating Data Cleaning

The rating and rating count columns contained non-numeric values. These columns were converted into numeric format using Pandas with error handling.

Rows that resulted in missing values after conversion were removed.  
An explicit copy of the DataFrame was created to avoid chained assignment warnings and ensure reliable preprocessing.


## ❓ Business Question 1  
### What is the overall sales trend?

The dataset does not contain order dates or time-based information. Therefore, traditional time-series sales trend analysis is not possible.

To address this limitation, the overall sales trend is analyzed using product pricing and discount patterns. By examining the distribution of actual prices, discounted prices, and discount percentages, we can understand how sales are positioned across different price ranges and discount levels.

This approach provides insight into overall sales behavior based on pricing strategies.


```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.histplot(
    df['discounted_price'],
    bins=30,
    kde=True,
    color='teal'
)

plt.title("Distribution of Discounted Prices", fontsize=12)
plt.xlabel("Discounted Price")
plt.ylabel("Number of Products")

# Save the figure
plt.savefig("overall_sales_trend_discounted_price.png", dpi=300, bbox_inches='tight')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df[['actual_price', 'discounted_price']],
    palette=['orange', 'green']
)

plt.title("Actual Price vs Discounted Price", fontsize=12)
plt.ylabel("Price")

# Save the figure
plt.savefig("actual_vs_discounted_price.png", dpi=300, bbox_inches='tight')

plt.show()
```

### Overall Sales Trend – Insights

- Most products are priced within a lower to mid-price range after discounts.
- There is a noticeable difference between actual and discounted prices, indicating active discount strategies.
- Discounted pricing plays a significant role in shaping overall sales behavior on the platform.

Although time-based trends are unavailable, pricing distribution provides a clear picture of overall sales positioning.


## ❓ Business Question 2  
### Which are the Top 10 products by sales?

The dataset does not contain direct information about quantity sold or total revenue.  
Therefore, sales are **estimated** using the following logic:

**Estimated Sales Value = Discounted Price × Rating Count**

This estimation assumes that products with higher ratings count indicate higher customer purchases, making it a reasonable proxy for sales performance.

Using this approach, the Top 10 products by estimated sales value are identified.


```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# Create estimated sales column
df.loc[:, 'estimated_sales'] = df['discounted_price'] * df['rating_count']

# Get Top 10 products
top_10 = (
    df.groupby('product_name')['estimated_sales']
    .sum()
    .sort_values(ascending=True)
    .tail(10)
)

# Shorten product names for readability
short_names = [name[:45] + "..." if len(name) > 45 else name for name in top_10.index]

plt.figure(figsize=(12,6))

plt.barh(short_names, top_10.values, color='purple')

plt.title("Top 10 Products by Estimated Sales Value", fontsize=13)
plt.xlabel("Estimated Sales Value")
plt.ylabel("Product Name")

# Format x-axis (no scientific notation)
plt.gca().xaxis.set_major_formatter(
    ticker.FuncFormatter(lambda x, pos: f"{int(x/1e6)}M")
)

# Save polished figure
plt.savefig(
    "top_10_products_by_estimated_sales_final.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

### Insights – Top 10 Products by Sales

- A small number of products contribute significantly to the overall estimated sales.
- Products with high customer engagement (rating count) and competitive pricing dominate the top positions.
- These products can be considered key revenue drivers for the platform.

Focusing on inventory availability and promotions for these products may improve overall sales performance.

## ❓ Business Question 5  
### Which are the most profitable category and sub-category?

The dataset does not contain a direct profit column.  
Therefore, profitability is **approximated** using product pricing and customer engagement.

For this analysis:
- Higher **average discounted price**
- Higher **customer ratings**
are considered indicators of better profitability.

Using this approach, category and sub-category performance is evaluated.


```python
# Split category into main category and sub-category
df[['main_category', 'sub_category']] = df['category'].str.split('|', expand=True)[[0, 1]]
```

```python
# Average discounted price by main category
category_profit = (
    df.groupby('main_category')['discounted_price']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))

category_profit.plot(
    kind='bar',
    color='coral'
)

plt.title("Top Categories by Average Discounted Price", fontsize=13)
plt.xlabel("Category")
plt.ylabel("Average Discounted Price")

plt.savefig(
    "top_profitable_categories.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
# Average discounted price by sub-category
sub_category_profit = (
    df.groupby('sub_category')['discounted_price']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))

sub_category_profit.plot(
    kind='bar',
    color='seagreen'
)

plt.title("Top Sub-Categories by Average Discounted Price", fontsize=13)
plt.xlabel("Sub-Category")
plt.ylabel("Average Discounted Price")

plt.savefig(
    "top_profitable_sub_categories.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

### Insights – Category and Sub-Category Profitability

- Categories with higher average discounted prices tend to generate higher revenue per product.
- Electronics and appliance-related categories dominate the top positions.
- Certain sub-categories consistently show higher pricing, indicating premium product segments.

Focusing on these categories can help improve overall profitability.
